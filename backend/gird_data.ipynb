# Grid Data rules

```python
def check_neighbors(r, c):
    val = grid[r][c]["value"]
    if val is None:
        return "Cell is empty"
        
    #Check right cell
    if c + 1 < 3: # don't fall off the edge
        right_neighbor = grid[r][c+1]["value"]
        if right_neighbor == val:
            return "Same number cells must not touch"
    
    return "YAASSS ZADDY"
```

# Implement Placing a Number in Cell (Basic Move)

```python
def make_move(grid, fixed_cells, row, col, value):
    """Place a number in a Fillomino puzzle cell.
    
    Args:
        grid: 2D list representing puzzle board
        fixed_cells: set of (row, col) tuples that are givens
        row, col: target cell coordinates
        value: number to place
        Returns:
            dict with "status" and "message"
        """
    # Input validation
    if not (0 <= row < len(grid) and 0 <= col < len(grid[0])):
        return {"status": "ERROR", "message": "Cell coordinates out of bounds"}
    
    if not (1 <= value <= 9):  # Adjust range if needed
        return {"status": "ERROR", "message": "Value must be between 1 and 9"}
    
    # Reject if cell is a "given"
    if (row, col) in fixed_cells:
        return {"status": "ERROR", "message": "Cannot modify given cells"}
    
    # if cell already has same number
    if grid[row][col] == value:
        return {"status": "SUCCESS", "message": "No change needed"}
    
    # Update the cell
    grid[row][col] = value
    return {"status": "SUCCESS","message": "Cell updated"}
```

```python
board = [[5, 0], [0, 0]]
givens = {(0, 0)}

# Your existing tests
print(make_move(board, givens, 0, 0, 3))   # ERROR: given cell
print(make_move(board, givens, 1, 1, 3))   # SUCCESS: updated
print(make_move(board, givens, 1, 1, 3))   # SUCCESS: no change

# New validation tests
print(make_move(board, givens, 5, 5, 3))   # ERROR: out of bounds
print(make_move(board, givens, 1, 1, 0))   # ERROR: invalid value
print(make_move(board, givens, 1, 1, 10))  # ERROR: invalid value
```
