# Test grid size

```python
# 1. Create a row template
def make_row():
    return [{"value": None, "is_given": False}, 
            {"value": None, "is_given": False}, 
            {"value": None, "is_given": False}]

# Stack rows into a grid
grid = [make_row(), make_row(), make_row()]

# Add clues
grid[0][0]["value"] = 2
grid[0][0]["is_given"] = True

grid[1][2]["value"] = 1                                                      
grid[1][2]["is_given"] = True
```

```python
for row in grid:
    print(row)
```

# Grid Data rules

```python
def check_neighbors(r, c):
    val = grid[r][c]["value"]
    if val is None:
        return "Cell is empty"
        
    #Check right cell
    if c + 1 < 3: # don't fall off the edge
        right_neighbor = grid[r][c+1]["value"]
        if right_neighbor == val:
            return "Same number cells must not touch"
    
    return "YAASSS ZADDY"
```

```python
print(check_neighbors(0,0))
```

# Implement Placing a Number in Cell (Basic Move)

```python
def make_move(grid, fixed_cells, row, col, value):
    # 1. Reject if cell is a "given"
    if (row, col) in fixed_cells:
        return {"status": "ERROR", "message": "Cannot modify given cells"}

    # 2. if cell already has same number, no change needed
    if grid[row][col] == value:
        return {"status": "SUCCESS", "message": "No change needed"}

    # 3. Update the cell(new or existing)
    grid[row][col] = value
    return {"status": "SUCCESS", "message": "Cell updated"}
```

```python
# Initial state: (0,0) is a fixed '5'. (1,1) is empty (0).
board = [[5, 0], [0, 0]]
givens = {(0, 0)}

print(make_move(board, givens, 0, 0, 3)) # Should return ERROR
print(make_move(board, givens, 1, 1, 3)) # Should return SUCCESS
print(make_move(board, givens, 1, 1, 3)) # Should return SUCCESS (No change)
```
